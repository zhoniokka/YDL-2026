## Titanic dataset — basic exploration

Load the Titanic dataset from seaborn and explore it with basic pandas functions.

```python
import numpy as np
import pandas as pd
import seaborn as sns

# Load the Titanic dataset
df = sns.load_dataset("titanic")
```

```python
# First 5 rows
df.head()
```

```python
# Column dtypes, non-null counts, and memory usage
df.info()
```

```python
# Shape: (rows, columns)
print("Shape:", df.shape)

# Summary statistics for numeric columns
df.describe()
```

```python
# Survival rate = mean of the 0/1 'survived' column, grouped by sex
survival_by_sex = df.groupby("sex")["survived"].mean()
survival_by_sex
```

```python
summary = df.groupby("sex")["survived"].agg(
    total="count",
    survived="sum",
)
summary["died"] = summary["total"] - summary["survived"]
summary["survival_rate_%"] = (summary["survived"] / summary["total"] * 100).round(1)
summary
```
